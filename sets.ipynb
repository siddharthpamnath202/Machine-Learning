```python
# Creating the sets 

my_set = {1,2,3,4,5}
print(my_set)
```

```python
my_set = set([1,2,2,4,5,6,7])
print(my_set)
```

```python
# Basic set operation 
# Adding and removing elements 
my_set.add(0)
my_set
```

```python
my_set.remove(1)
my_set
```

```python
my_set.discard(11)
my_set
```

```python
my_set.pop()
my_set
```

```python
# clear the elemnts 
my_set.clear()
my_set
```

```python
# Set membership test 
my_set = (1,2,3,4,5)
print(3 in my_set)
print(10 in my_set)
```

```python
# Mathemical opeartions 
set1 = {1,2,3,45,6,7}
set2 = {6,4,6,7,7,8}

# Union 
union_set =set1.union(set2)
union_set

# Intersection
intersection_set = set1.intersection(set2)
intersection_set

set1.intersection_update(set2)

set1
```

```python
set1 = {1,2,3,45,6,7}
set2 = {6,4,6,7,7,8}

# Difference
set1.difference(set2)
```

```python
# Symmetric Difference 
set1.symmetric_difference(set2)
```

```python
# sets methods 
set1={1,2,3}
set2={3,4,5}

# What is subset 
print(set1.issubset(set2))
```
