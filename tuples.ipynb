```python
# Immutable nature of tuples 
# Tupes are immutable, meaning thier elements cannot be changged once assigned

lst = [1,2,3,4,5]
print(lst)

lst[1] = "krishna"
print(lst)
```

```python
numbers[1] = "Siddharth"
```

```python
# Tuple methods 
print(numbers.count(1))
print(numbers.index(3))
numbers
```

```python
# Packing and unpacking 
# 
packed_tuple = 1,"Hello",3.12
print(packed_tuple)
```

```python
# Unacking Tuple 
a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
# Unpacking with * 
numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(type(middle))
print(middle)
print(last)
```
