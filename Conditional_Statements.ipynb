```python
# Leap Year 

# Determine if a year is a leap number nested condition statement 

year = int(input("Enter the year"))
if year%2==0 :
    if year%100 == 0:
        if year%400 == 0:
            print("This is leap year",year)
        else:
            print("This is not a leap year",year)
    else : 
        print("Year is the leap year ",year)
else:
    print("Year is not a leap year",year)
```
