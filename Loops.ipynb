```python
# While loop 
# Continues excutes as long as the condition is true 
count = 0

while count <= 5 :
    print(count)
    count = count+1
```

```python
count = 0



while count%2 == 0:
    print("Count is",count)
    count = count + 1
```

```python
# Loop control statments 

# Break
# The break statment exits the loop prematurely
# Break 

for i in range(10):
    if i == 5 :
        print("Breaked")
        break
    print(i)

```

```python
# Continue statment 
#  The continue statment skips the current iteration and continues with the next 

for i in range(10):
    if i%2 == 0:
        continue    
    print(i)
```

```python
# Pass - statemeent is the null operation it does nothing 

for i in range (5):
    if i == 3 :
        print("The number is i ",i)
        pass
    print(i)
```

```python
# Nested Loops
# A llop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j : {j}")
```

```python
# Example : Calculte the sm of the first n natural numbers using the while and for loop 

num = 5
sum = 0 
count  = 1 
while count <=num:
    sum = sum +count
    count = count +1 
print("Sum of first natural numbers ", sum)
```
