Dictionaries - : Unordered collection of items , They store data in the key value pairs . Keys must be unique and immutable (eg : strings,number or tuples) while values can be of any type

```python
# Single key no duplicate key should be present

student = {
    "surname" : "Pamnath",
    "name" : "siddharth",
    "age" : 23,
}
print(student)
print(type(student))
```

```python
# Iterating over Dictionaries 
# You can use loops to iterate over dictionaries , keys , values or items 
# Iterating over keys 

for i in student.keys():
    print(i)
```

```python
for j in student.values():
    print(j)
```

```python
# Iterate  over key value pairs 
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
#  Nested Dictionaries

student = {
    "student1" : {"name" : "siddharth","age" : 32},
    "student2" : {"name" : "Pooja" ,"age": 99 } 
}

student
```

```python
# Acess nested dictionaries elements 
print(student["student2"]["name"])
print(student["student1"]["age"])
student.items()
```

```python
# Iterating over nested dictionraies 

for student_id, student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}{value}")
```
